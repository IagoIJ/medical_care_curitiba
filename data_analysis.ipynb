```python
df_clean['cd_CID_resumido'] = df_clean['Código do CID'].str.strip().str[:3]
```

```python
#Clusterizar paciêntes para enteder melhor o perfil dos pacientes que estão sendo atendidos no sistema de saúde.

columns_cluster_paciente = [
    'Sexo','cd_CID_resumido','Energia Elétrica','Tipo de Habitação','Destino Lixo', 'Fezes/Urina', 'Cômodos','Em Caso de Doença','Meio de Comunicacao','Meio de Transporte','Tratamento no Domicílio'
]

df_cluster_paciente = df_clean[columns_cluster_paciente]
```

```python
print(df_cluster_paciente.shape)

print(df_cluster_paciente.dropna().shape)
```

```python
X = df_cluster_paciente.dropna()
```

```python
import pandas as pd
import numpy as np

cid = (
    X['cd_CID_resumido']
    .astype('string')
    .str.strip()
    .str.upper()
)

letra = cid.str[0]

numero = pd.to_numeric(
    cid.str.slice(1, 3),
    errors='coerce'
)

condicoes = [
    letra.isin(['A', 'B']),
    (letra == 'C') | ((letra == 'D') & (numero <= 48)),
    (letra == 'D') & (numero >= 50),
    letra == 'E',
    letra == 'F',
    letra == 'G',
    (letra == 'H') & (numero <= 59),
    (letra == 'H') & (numero >= 60),
    letra == 'I',
    letra == 'J',
    letra == 'K',
    letra == 'L',
    letra == 'M',
    letra == 'N',
    letra == 'O',
    letra == 'P',
    letra == 'Q',
    letra == 'R',
    letra.isin(['S', 'T']),
    letra.isin(['V', 'W', 'X', 'Y']),
    letra == 'Z',
    letra == 'U'
]

capitulos = np.arange(1, 23)

X['grupo_cid'] = np.select(
    condicoes,
    capitulos,
    default=0
).astype(np.int8)
```

```python
X = X.drop(columns=['cd_CID_resumido'])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.cluster import KMeans

colunas_numericas = ['Cômodos']


colunas_categoricas = X.drop(columns=colunas_numericas).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            colunas_numericas
        ),
        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=True
            ),
            colunas_categoricas
        )
    ]
)

X_processed = preprocessor.fit_transform(X)
```
